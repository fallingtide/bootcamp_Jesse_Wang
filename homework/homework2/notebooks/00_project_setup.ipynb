# Environment & Config Check

Verifies that the Stage 02 tooling is set up correctly:

- `.env` is read via `python-dotenv` (`API_KEY` should be present)
- `src/config.py` loads and reads config values
- NumPy is importable and runs a small array operation

```python
import os
import sys
from pathlib import Path

import numpy as np
from dotenv import load_dotenv

# Make the project's src/ package importable regardless of launch directory
root = Path.cwd()
project_root = root if (root / "src").is_dir() else root.parent
sys.path.insert(0, str(project_root))

# Load secrets/config from .env
load_dotenv(project_root / ".env")

# 1) Config helper check — does .env get read correctly?
from src.config import get_key

api_key = get_key("API_KEY")
print("API_KEY present:", bool(api_key))
print("DATA_DIR:", get_key("DATA_DIR"))

# 2) NumPy demo
arr = np.arange(5)
print("numpy version:", np.__version__)
print("arr:", arr)
print("arr * 2:", arr * 2)
```
